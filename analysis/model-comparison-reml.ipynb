# Model Comparison Using REML and Effective Degrees of Freedom

Compare four candidate models using REML score and model complexity.

## Data and reproducibility

The participant-level NIPT dataset is not distributed in this public repository. To reproduce the analysis, place an authorized copy at `data/nipt_data.xlsx` using the English schema documented in `data/README.md`.


## Visualize the model comparison


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams


rcParams["font.family"] = ["DejaVu Sans"]
rcParams["axes.unicode_minus"] = False

# data
models = ["m1", "m2", "m3", "m4"]
edf = np.array([248.68, 248.14, 255.19, 255.35])
reml = np.array([-2269.958, -2270.54, -2284.516, -2279.608])

# Efficiency frontier
order = np.argsort(edf)
edf_sorted, reml_sorted = edf[order], reml[order]
cum_min = np.minimum.accumulate(reml_sorted)
front_x = edf_sorted[reml_sorted <= cum_min + 1e-12]
front_y = reml_sorted[reml_sorted <= cum_min + 1e-12]


plt.figure(figsize=(7.2, 5))

for x, y, m in zip(edf, reml, models):
    if m == "m3":
        plt.scatter(x, y, s=150, color="tab:blue", edgecolor="black", zorder=3, label="m3 (lowest REML)")
    else:
        plt.scatter(x, y, s=70, color="gray", alpha=0.7, zorder=2)
    plt.annotate(m, (x, y), textcoords="offset points", xytext=(0, 6),
                 ha="center", fontsize=10)

plt.plot(front_x, front_y, linestyle="--", linewidth=1.5, color="tab:blue")

m3_idx = models.index("m3")
plt.annotate("m3 selected model",
             xy=(edf[m3_idx], reml[m3_idx]),
             xytext=(edf[m3_idx]-2, reml[m3_idx]+7),
             arrowprops=dict(arrowstyle="->", lw=1.0),
             ha="right", fontsize=11)

plt.title("Model selection: REML versus EDF", fontsize=13)
plt.xlabel("Total effective degrees of freedom (lower is simpler)", fontsize=11)
plt.ylabel("REML score = 2 x log-likelihood (lower is better)", fontsize=11)


plt.grid(False)

plt.tight_layout()
plt.show()
```
